```r
# Plotting the Temperature Data (The built-in dataset nhtemp contains average annual temperatures (presumably for New Haven))
## type="p" tells R to plot points.
## pch=20 specifies a solid circle for each data point.
## lwd=3 increases the line width for the points (though it mostly affects lines, it can alter the point border thickness).
plot(nhtemp, type="p", ylab="temp", pch=20, lwd=3)
```

```r
time = rownames(nhtemp)
plot(nhtemp, type="p", ylab="temp")
# col=2 (typically red) and lwd=2 set the color and line width of the regression line
# lm(nhtemp ~ seq(1912,1971, by=1)):
#
## The function call seq(1912,1971, by=1) generates a sequence of numbers starting at 1912 and ending at 1971, 
## with each number increasing by 1. Essentially, it creates a vector that looks like:
## 1912, 1913, 1914, ..., 1971
## This is called: independent variable (or "time" variable)
#
## nhtemp is the observed data, the observed temperatures, typically a numeric vector containing temperature data over a series of years
#
## "nhtemp ~ seq(1912,1971, by=1)" tells R to model the temperature data as a linear function of the year
abline(lm(nhtemp ~ seq(1912,1971, by=1)), col=2, lwd=2)

## Residuals:
### The segments drawn between the predicted and actual values illustrate the deviations (errors) of the model, 
### helping assess the model’s fit:
segments(seq(1912,1971, by=1), predict(lm(nhtemp ~ seq(1912,1971, by=1))), seq(1912,1971, by=1), nhtemp)
```
