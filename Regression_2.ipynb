```r
# The trees dataset is a built-in dataset in R. It's included in the base R distribution via the datasets package, 
# which is loaded automatically. 
# This dataset provides measurements for 31 trees, including variables like Girth, Height, and Volume

# lm builds the model out of the two vectors of data: "trees$Volume" and "trees$Girth"
# The model is designed to investigate how the logarithm of Volume (the dependent variable) is related to the logarithm of 
# Girth (the independent variable) in the trees dataset
fm1 <- lm(log(Volume) ~ log(Girth), data = trees)
# summary(fm1)

# First we plot the scatter points:
# We use log(trees$Girth) and log(trees$Volume) because the regression model was built on the logarithms of these variables
plot(log(trees$Girth), log(trees$Volume), 
     type = "p", # specifies that we’re plotting points
     xlab = "log(Girth)", # Labelling the x-axis
     ylab = "log(Volume)", # Labelling the y-axis
     pch = 20, # sets the plotting symbol to a solid circle
     lwd = 3) # increases the thickness of the plotted symbols (or their borders).

# Secondly, we plot the ab line:
# This adds the regression line obtained from the model fm1 onto the scatterplot
abline(fm1, col = 2, lwd = 2) # col = 2 (red color), lwd = 2 (line width of 2)
```
